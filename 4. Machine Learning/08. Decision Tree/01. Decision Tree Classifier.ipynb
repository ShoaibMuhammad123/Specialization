## **DECISION TREE CLASSIFIER IMPLIMENTATION**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
iris.data[:10]
```

```python
# to check the description of the data
print(iris.DESCR)
```

```python
print(iris.target)
```

```python
column_names = iris.feature_names
column_names
```

#### **Indepandent and depandent features**

```python
## Independent features
X = pd.DataFrame(iris.data,columns=column_names)
```

```python
X.head()
```

```python
## dependent features
y = iris['target']
```

```python
y[:5]
```

#### Train test split

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2,random_state=10)
```

### **Applying Decision Tree Classifier**

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier()
```

```python
dt.fit(X_train,y_train)
```

## **Visualizing Decision Tree**

```python
from sklearn import tree
plt.figure(figsize=(15,12))
tree.plot_tree(dt,filled=True)
plt.show()
```

* `This is how the split can happen`
* `One problem can occur over here is Overfitting`

## `Preventing Overfitting`
* To prevent overfitting we can use two techniques over here.
1. `Post pruning` (like we are saying after max depth 2 no furthur splitting require etc. )
2. `Pre Pruning`

## `Post Pruning`

### If we apply max-depth then what happen see below

```python
dt = DecisionTreeClassifier(max_depth=3)
dt.fit(X_train,y_train)
plt.figure(figsize=(15,12))
tree.plot_tree(dt,filled=True)
plt.show()
```

```python
## to make prediction 
y_pred = dt.predict(X_test)
```

```python
# evaluation
from sklearn.metrics import classification_report, confusion_matrix
```

```python
cm = confusion_matrix(y_test,y_pred)
print(cm)
print(classification_report(y_test,y_pred))
```

`Post pruning` is alwayas good for smaller dataset
`Pre pruning` is best more we use this
