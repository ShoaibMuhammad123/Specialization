## **Hierarichal Clustering Implementation**

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn import datasets
```

```python
## Importing Iris dataset 
iris = datasets.load_iris()
```

```python
iris_data = pd.DataFrame(iris.data)
```

```python
iris_data.columns = iris.feature_names
```

```python
iris_data
```

### In order to perform clustering first we do **Scaling**

```python
from sklearn.preprocessing import StandardScaler
```

```python
## Standardization

scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(iris_data)
```

```python
X_scaled
```

## Now we applying PCA to **reduce the number of Dimention**

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
pca_scaled_data = pca.fit_transform(X_scaled)
```

```python
pca_scaled_data
```

```python
### now we plot the data points 
plt.scatter(pca_scaled_data[:,0],pca_scaled_data[:,1],c=iris.target)
plt.show()
```

### These are the **three classes in our data**

## Applying `Agglomerative Clustering`

* `In Agglomerative Clustering the first step is to construct a **dendogram**`
* `To plot a dendogram we use scipy library`

```python
pip install scipy
```

```python
import scipy.cluster.hierarchy as sc
## plot the dendogram
plt.figure(figsize=(20,7))
plt.title('Dendograms')

## create dendogram
dendrogram = sc.dendrogram(sc.linkage(pca_scaled_data, method='ward'))
plt.title('Dendogram')
plt.xlabel('Sample Index')
plt.show()
```

* `The below longest vertical line have non horizontal line passing except the one we draw over it` 
* `The number of cluster is 2 because our horizontal line passes in two line , so it has just 2 points`

### Now we apply Agglomerative Clustering

```python
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters=2,metric ='euclidean',linkage='ward')
cluster.fit(pca_scaled_data)
```

```python
cluster.labels_
```

```python
### now we plot the data points 
plt.scatter(pca_scaled_data[:,0],pca_scaled_data[:,1],c=cluster.labels_)
plt.show()
```

### `This is is how Hierarichal Clustering (Agglomerative Clustering) applied`

* The main thing over here is creating dendogram.
