### **Stokes Prediction Continuation with SVC**

```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('../04. Logistic Regression\practice\Clean_Stokes_Data.csv')
```

```python
df.head()
```

```python
df['work_type'].value_counts()
```

```python
df['smoking_status'].value_counts()
```

```python
## Label Encoding 
df['gender'] = df['gender'].map({'Male':1,'Female':0})
```

```python
df['ever_married'] = df['ever_married'].map({'Yes':1,'No':0})
df['Residence_type'] = df['Residence_type'].map({'Urban':1,'Rural':0})
```

## **Nominal Encoding with OneHotEncoding**

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder()

encoded = ohe.fit_transform(df[['work_type','smoking_status']]).toarray()
encoded
```

```python
encoded_df = pd.DataFrame(encoded,columns=ohe.get_feature_names_out())
encoded_df.head()
```

```python
## concatinating original df and encoded df 
final_df  = pd.concat([df,encoded_df],axis=1)
final_df.head()
```

```python
## droping the two nominal columns  smoking_status and work_type
final_df.drop(columns=['work_type','smoking_status'],axis=1,inplace=True)
```

```python
final_df.shape
```

```python
final_df.sample(3)
```

```python
final_df.isnull().sum()
```

```python
final_df = final_df[~final_df['gender'].isna()]
```

```python
final_df.isnull().sum()
```

#### Now we will Scale some columns 

```python
X = final_df.drop('stroke',axis=1)
y = final_df['stroke']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
```

```python
X_train.head(2)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train[['age','avg_glucose_level','bmi']])
X_test_scaled = scaler.transform(X_test[['age','avg_glucose_level','bmi']])
```

```python
X_train_scaled
```

```python
X_test_scaled
```

```python
X_train[['age','avg_glucose_level','bmi']] = X_train_scaled
X_test[['age','avg_glucose_level','bmi']] = X_test_scaled
```

```python
X_train.head()
```

```python
X_test.head()
```

## Now we **apply SVC**

```python
from sklearn.svm import SVC
```

```python
svc = SVC(kernel='rbf')
svc.fit(X_train,y_train)
y_prediction = svc.predict(X_test)
```

```python
y_prediction
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,precision_score,recall_score,f1_score
import warnings
warnings.filterwarnings('ignore')
```

```python
print('Confusion Matrix: \n',confusion_matrix(y_test,y_prediction))
```

```python
y_test.value_counts()
```

```python
print(classification_report(y_test,y_prediction))
```

### Class Weights 

```python
svc = SVC(kernel='rbf',class_weight={0:1,1:20})
svc.fit(X_train,y_train)

y_prediction = svc.predict(X_test)

print('Confusion Matrix: \n',confusion_matrix(y_test,y_prediction))
print(classification_report(y_test,y_prediction))
```

## **Upsampling and Downsampling to Improve the Result**

```python
from imblearn.over_sampling import SMOTE
```

```python
smote = SMOTE(random_state=42)
```

```python
X_train_oversample,y_train_oversample = smote.fit_resample(X_train,y_train)
```

```python
X_train_oversample
```

```python
X_train.shape,X_train_oversample.shape
```

```python
y_train_oversample.value_counts()
```

### **SVC on Oversampled Data**

```python
## for kernel = 'linear'
from sklearn.svm import SVC
svc = SVC(kernel='linear')
svc.fit(X_train_oversample,y_train_oversample)
y_pred = svc.predict(X_test)

print('Confusion Matrix: \n',confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
# for kernel = 'poly'

svc = SVC(kernel='poly')
svc.fit(X_train_oversample,y_train_oversample)
y_pred = svc.predict(X_test)

print('Confusion Matrix: \n',confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
# for kernel = 'rbf'

svc = SVC(kernel='rbf')
svc.fit(X_train_oversample,y_train_oversample)
y_pred = svc.predict(X_test)

print('Confusion Matrix: \n',confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
# for kernel = 'sigmoid'

svc = SVC(kernel='sigmoid')
svc.fit(X_train_oversample,y_train_oversample)
y_pred = svc.predict(X_test)

print('Confusion Matrix: \n',confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

* `This show that applying oversampling has imporve result for class 1 (patient having strokes)`
* `Specially for kernel = 'linear' for class 1, it is gving high result but overall for the other class it is making alot of error`

## **Downsampling**

```python
X_train['stroke'] = y_train
```

```python
X_train['stroke'].value_counts()
```

```python
df_minority = X_train[X_train['stroke']==1]
df_majority = X_train[X_train['stroke']==0]

```

```python
df_majority.shape,df_minority.shape
```

```python
from sklearn.utils import resample
df_majority_downsampled = resample(df_majority,
                                    replace=False,
                                    n_samples=len(df_minority),
                                    random_state=42,
                                    )
```

```python
### Concatinate Majority downsampled and Minorty downsampled
downsampled_df = pd.concat([df_majority_downsampled,df_minority],axis=0)
```

```python
downsampled_df.shape
```

```python
downsampled_df.head()
```

```python
downsampled_df = downsampled_df.sample(frac=1)   ## Shuffling the dataframe 
downsampled_df.head()
```

```python
X_train_downsampled = downsampled_df.iloc[:,:-1]
y_train_downsampled = downsampled_df.iloc[:,-1]
```

```python
X_train_downsampled.shape
```

```python
y_train_downsampled.shape
```

### **Training SVM on Downsampled Data**

```python
svc = SVC(kernel='linear')
svc.fit(X_train_downsampled,y_train_downsampled)
y_pred = svc.predict(X_test)

print('Confusion Matrix: \n',confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
svc = SVC(kernel='sigmoid')
svc.fit(X_train_downsampled,y_train_downsampled)
y_pred = svc.predict(X_test)

print('Confusion Matrix: \n',confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
svc = SVC(kernel='rbf')
svc.fit(X_train_downsampled,y_train_downsampled)
y_pred = svc.predict(X_test)

print('Confusion Matrix: \n',confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
svc = SVC(kernel='poly')
svc.fit(X_train_downsampled,y_train_downsampled)
y_pred = svc.predict(X_test)

print('Confusion Matrix: \n',confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
svc = SVC(kernel='poly')
svc.fit(X_train_downsampled,y_train_downsampled)
y_pred = svc.predict(X_test)

print('Confusion Matrix: \n',confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
