## Applying Principle Component Analysis

*  First we Standardize the data so, that the data mean come to 0 and standard deviation come to 1, all the valuees usually lies b/w -3 to +3

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaler.fit(df)
```

```python
scaled_data = scaler.transform(df)
```

```python
scaled_data
```

* Now i want `out of all features` i want to `extract only 2 features`.

```python
## for this we apply PCA Algorithm
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)   # n_components =2 mean i want 2 features of of all the feautres
```

```python
data_pca = pca.fit_transform(scaled_data)
```

```python
## This is our transformed data

data_pca   
```

* pca convert the data in such a way that `maximum variance` is getting captured through eigen decomposition concepts.

#### This process is known as `dimentionality reduction`

```python
### Now we can plot this data to see the distribution
pca.explained_variance_
```

* This is the variance of `feature1` and `feature2`. 

### Note:
* If we apply without n_component then it create the total number of features which contain in your dataset. i.e if our data contain 7 feature then after pca it also gives us 7 features.

```python
## now we see what we can do with the help of these two features
plt.figure(figsize=(8,6))
plt.scatter(data_pca[:,0],data_pca[:,1],c=cancer_dataset['target'],cmap='plasma')
plt.xlabel('First principal Component')
plt.ylabel('Second Principle Component')
plt.show()
```

* Alwasys remember `1st principle component` captured `maximum variance` as compared to `2nd principle component`

### 

```python
## if you want to select 3 features after pca then you just need to select n_component=3

pca = PCA(n_components=3)
```

```python
pca.fit_transform(scaled_data)
```

#### These are the three features which you require.
