# **KMean Clustering Algorithm Implimentation**

```python
import pandas as pd
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
```

```python
X,y = make_blobs(n_samples=1000,
           n_features=2,
           centers=3,)
```

```python
X.shape,y.shape
```

* `Because we are using Unsupervised Machine Learning Algorithm so we would not use the labels for training`
* `Centers are represent the total number of groups`

```python
## now we plot to see the distribution of data
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
## Now we perform Standardization
from sklearn.preprocessing import StandardScaler
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2,random_state=42)
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
plt.scatter(X_train_scaled[:,0],X_train_scaled[:,1])
plt.title('After Scaling')
```

### Applying KMeans Algorithm

```python
### wcss---> with cluster sum of squared
wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters=k,init='k-means++')
    kmeans.fit(X_train_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

## `How to determine what should be the k value`

```python
## in order to determine the exact k value we plot the wcss
```

### Plot Elbow Curve

```python
plt.plot(range(1,11),wcss)
plt.xticks(range(1,11))
plt.title('Elbow Curve')
plt.xlabel('Number of Clusters')
plt.ylabel('ylabel')
plt.show()
```

```python
kmeans = KMeans(n_clusters=3,init='k-means++')
```

```python
kmeans
```

```python
kmeans.fit_predict(X_train_scaled)
```

```python
## prediction on test data
y_pred = kmeans.predict(X_test_scaled)
```

```python
y_pred
```

```python
## now we plot to see the distribution of data
plt.scatter(X_test[:,0],X_test[:,1],c=y_pred)
```

###  `Silhoutee Scoring`

```python
from sklearn.metrics import silhouette_score
```

```python
silhouette_coefficients=[]
for k in range(2,11):
    kmeans = KMeans(n_clusters=k,init='k-means++')
    kmeans.fit(X_train_scaled)
    score = silhouette_score(X_train_scaled,kmeans.labels_)
    silhouette_coefficients.append(score)
```

```python
silhouette_coefficients
```

```python
plt.plot(range(2,11),silhouette_coefficients)
plt.xticks(range(1,11))
plt.title('Elbow Curve')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhoutte coefficient')
plt.show()
```

## Silhoutte coefficient max value will be consider as k value , here it has 3 which is a maximum value
