# **Cleaning the Dataset**

```python
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header=1)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
## row wise null values 
df[df.isnull().any(axis=1)]
```

```python
## Index 122 has been removed 

df.drop(index=122,inplace=True)
```

```python
df.reset_index(drop=True,inplace=True)
```

```python
df.iloc[[122]]
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df.iloc[[122]]
```

```python
df = df[df.index!=122]
```

```python
df.shape
```

```python
df.reset_index(drop=True,inplace=True)
```

```python
df.shape
```

```python
df.info()
```

```python
df.sample()
```

```python
df.columns
```

```python
df.columns = df.columns.str.strip()
```

```python
df.columns
```

```python
df.head()
```

```python
df[['day','month','year','Temperature','RH','Ws']] = df[['day','month','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

```python
## converting the objects in to float 
objects = [feature for feature in df.columns if df[feature].dtype=='object']
```

```python
objects
```

```python
## conversting these objects in to float 
for i in objects:
    if i!='Classes':
        df[i] = df[i].astype(float)
```

```python
df.info()
```

```python
df.head()
```

```python
## saving the clean data 
df.to_csv('Algerian_forest_fires_cleaned_dataset.csv',index=False)
```

# **Exploratory Data Analysis**

```python
df=pd.read_csv('Algerian_forest_fires_cleaned_dataset.csv')
```

```python
df.head()
```

```python
### day month and year is not that much important for my case so we remove them 

df_copy = df.drop(columns=['day','month','year'],axis=1)
```

```python
df_copy.head()
```

```python
### encoding the categories in Classes 
df_copy['Classes'].value_counts()
```

```python
## Issue over here
df_copy['Classes'].value_counts().keys()
```

```python
df_copy['Classes'] = df_copy['Classes'].str.strip()
```

```python
df_copy['Classes'].value_counts()
```

```python
## Encoding the fire =1 and non fire = 0

df_copy['Classes'] = df_copy['Classes'].map({'fire':1,'not fire':0})
```

```python
df_copy['Classes'].value_counts()
```

```python
df_copy.head()
```

```python
## Show the percentage of Classes in graph form
percentage = df_copy['Classes'].value_counts(normalize=True)*100
percentage
```

```python
import seaborn as sns
percentage.plot(kind='pie',autopct ="%1.1f%%")
```

```python
## Plot density plot for all features 
# plt.style.use('seaborn')
df_copy.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
## Now we check correlation and multi collinearity
df_copy.corr()
```

```python
## to see the correlation in a better way we plot it 
plt.figure(figsize=(16,12))
sns.heatmap(df_copy.corr(),annot=True)
```

```python
## just checking correlation w r t target column fWI

plt.figure(figsize=(12,5))
df_copy.corr()['FWI'].plot()
plt.title('Correlation of all column w r t FWI Column')
plt.xlabel('All Indepandent Columns')
plt.ylabel('Correlation with FWI')
plt.show()
```

```python
df_copy.corr()['FWI']
```

```python
## To check the Outliers in target variable FWI
## box plot 
sns.boxplot(y=df_copy['FWI'],hue='Classes',data=df_copy)
```
