### **Applying Naive Bayes on Strokes Data**

```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings 
warnings.filterwarnings('ignore')
```

```python
train_df = pd.read_csv('../05. SVM\stroke_train_data.csv')
test_df = pd.read_csv('../05. SVM\stroke_test_data.csv')
```

```python
# df.drop(columns=['Unnamed: 0','index'],axis=1,inplace=True)
```

```python
## training data split
X_train,y_train = train_df.drop('stroke',axis=1),train_df['stroke']
```

```python
## test data split
X_test,y_test = test_df.drop('stroke',axis=1),test_df['stroke']
```

## `Applying Gussain Naive Bayes`

```python
from sklearn.naive_bayes import GaussianNB,BernoulliNB
```

```python
gnb = GaussianNB()
gnb.fit(X_train,y_train)
y_pred = gnb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
bnb = BernoulliNB()
bnb.fit(X_train,y_train)
y_pred = bnb.predict(X_test)
```

```python
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
