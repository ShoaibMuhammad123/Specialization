## Real Project

```python
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')

paragraphs = [
    """The product is very good. I am really enjoying
    using it. The quality is excellent and the delivery was fast.""",

    """The products were not working properly.
    I was disappointed because they arrived late.
    The company should improve its services.""",

    """I love shopping from this store.
    The products are affordable and the staff is very helpful.
    I will recommend this store to my friends."""
]
```

```python
for paragraph in paragraphs:

    words = word_tokenize(paragraph)

    print(words)
```

```python
for paragraph in paragraphs:

    words = word_tokenize(paragraph)

    pos_tags = nltk.pos_tag(words)

    print(pos_tags)
```

```python
stop_words = set(stopwords.words('english'))

for paragraph in paragraphs:

    words = word_tokenize(paragraph)

    filtered_words = []

    for word in words:
        if word.lower() not in stop_words:
            filtered_words.append(word)

    print(filtered_words)
```

```python
## stemming 
ps = PorterStemmer()

for paragraph in paragraphs:

    words = word_tokenize(paragraph)

    stemmed_words = [ps.stem(word) for word in words]

    print(stemmed_words)
```

```python
### lammitization

lemmatizer = WordNetLemmatizer()

for paragraph in paragraphs:

    words = word_tokenize(paragraph)

    lemmatized_words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]

    print(lemmatized_words)
```

```python
for paragraph in paragraphs:

    words = word_tokenize(paragraph)

    pos_tags = nltk.pos_tag(words)

    lemmatized_words = []

    for word, tag in pos_tags:

        if tag.startswith('V'):
            pos = 'v'
        elif tag.startswith('J'):
            pos = 'a'
        elif tag.startswith('R'):
            pos = 'r'
        else:
            pos = 'n'

        lemmatized_words.append(
            lemmatizer.lemmatize(word, pos=pos)
        )

    print(lemmatized_words)
```
