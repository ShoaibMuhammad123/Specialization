```python
import pandas as pd
```

```python
data = pd.read_csv("data/email_spam.csv")
```

```python
data = data[['text','type']]
```

```python
data
```

```python
import re
```

```python
from nltk.stem import WordNetLemmatizer
lametizer= WordNetLemmatizer()
```

```python
corpus =[]

for i in range(0,len(data['text'])):
    review = re.sub('[^a-zA-Z]',' ', data['text'][i])
    review = review.lower()
    review = review.split()
    review = [lametizer.lemmatize(word) for word in review]
    review = ' '.join(review)
    corpus.append(review)
    
    
    
    
```

```python
corpus
```

```python
## Every Time you don't need to do all the things 

# what you can do is to use simple_preprocess

from nltk.tokenize import sent_tokenize
from gensim.utils import simple_preprocess

words =[]

for sent in corpus:
    sent_tokens = sent_tokenize(sent)
    for sent in sent_tokens:
        words.append(simple_preprocess(sent))
```

```python
words
```

## Now we Train the Word2Vec model from Scratch

```python
import gensim
```

```python
model = gensim.models.Word2Vec(words,epochs=1000)
```

```python
model.vector_size
```

```python
model.corpus_count
```

```python
model.wv.similar_by_word('can')
```

```python
import numpy as np 

def avg_word2vec(doc):
    
    return np.mean([model.wv[word] for word in doc if word in model.wv.index_to_key])
```
