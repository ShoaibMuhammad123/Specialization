# **BAG of Word Impliementation and Text Classification**

```python
import pandas as pd
```

```python
data = pd.read_csv("data/email_spam.csv")
```

```python
data
```

```python
data.shape
```

```python
## remove the title column 
data = data[['text','type']]
```

```python
data
```

```python
import re 
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer 
ps = PorterStemmer()
```

```python
data['type'].value_counts()
```

```python
len(data)
```

```python
corpus = []

for i in range(0,len(data)):
    review = re.sub('[^a-zA-Z]',' ',data['text'][i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
    print(review)
    print()
    
    
```

```python
corpus
```

## Now we Create Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer

### for binary Bag of Words enable binary =True 
cv = CountVectorizer(max_features=2500,ngram_range=(1,2))
```

```python
cv
```

```python
X = cv.fit_transform(corpus).toarray()
```

```python
X
```

```python
## to see the vocabulary 
cv.vocabulary_
```

```python
## Output Feature 

y = pd.get_dummies(data['type'])
```

```python
y = y['spam'].map({True: 1, False: 0})
y
```

```python
### Train TEst Split 
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2,random_state=42)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_detect = MultinomialNB().fit(X_train,y_train)
```

```python
y_pred = spam_detect.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
print(accuracy_score(y_pred,y_test))
```

```python
print(classification_report(y_pred,y_test))
```
