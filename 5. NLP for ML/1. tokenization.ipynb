```python
import nltk
```

```python
from nltk.tokenize import sent_tokenize
```

```python
corpus = [
    "This is the first Sentence", 
    "Now is is the second sentence"
]

for sent in corpus:
    print(sent_tokenize(sent))
```

```python
### Word Tokenizer 
# converting sentence into words(token)

from nltk.tokenize import word_tokenize

for word in corpus:
    print(word_tokenize(word))
```

```python
# this is how tokenization work 
# 1- if we are converting a paragraph into token then the token will be a sentence , if we are tokeinzing a sentence then token will be a word
```
