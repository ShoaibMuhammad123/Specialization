# **Lammitization**

```python
from nltk.stem import WordNetLemmatizer
```

```python
import nltk
nltk.download('wordnet')
```

```python
lammitizer = WordNetLemmatizer()
```

```python
sentence =["finally","goes","running","went"]
```

```python
sentence
```

```python
for word in sentence:
    print(word,"--> ",lammitizer.lemmatize(word,pos="v"))
```

```python
for word in sentence:
    print(word,"--> ",lammitizer.lemmatize(word,pos="a"))
```

# This is how lammitization work
