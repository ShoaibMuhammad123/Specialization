# **Stemming**

`Stemming is the process of converting a word to its word stem.`

```python
from nltk.stem import PorterStemmer
```

```python
sentence = ["finally","going","went","goes","are","history"]
porterstemmer = PorterStemmer()
for word in sentence:
    print(word,"-->",porterstemmer.stem(word))
```

`This is how porter stemmer work , it give some words an incorrect root stem `

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = reg_stemmer = RegexpStemmer('ing$|es$|s$|',min=4)
```

```python
for word in sentence:
    print(word,"--->",reg_stemmer.stem(word))
```
