# **StopWords**

`Stop words are common words in a language that appear frequently in sentences but usually carry little meaningful information for text analysis.

In Natural Language Processing (NLP), stop words are often removed during text preprocessing to focus on important words.`

`Examples of stop words`
- Articles: a, an, the

- Prepositions: in, on, at, from, with

- Helping verbs: is, am, are, was, were

- Conjunctions: and, but, or

- Pronouns: he, she, it, they



```python
corpus ="""Yesterday, I went to the market with my friends to buy some fresh fruits and vegetables.
    We walked through the streets and visited a small shop near the park. The shopkeeper was 
    very friendly and offered us some discounts on different products. I bought apples, bananas, 
    and oranges for my family. After shopping, we went to a restaurant and ordered some delicious food.
    We enjoyed our meal and talked about our studies and future plans. It was a wonderful day,
    and I was very happy to spend time with my friends."""


corpus
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

stop_words = set(stopwords.words('english'))

words = word_tokenize(corpus)

filtered_words = []

for word in words:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print(filtered_words)
```

```python
clean_words = [ ''.join(word) for word in filtered_words if word not in [',','.',' '] ]
clean_words
```
