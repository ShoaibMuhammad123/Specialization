1. Write a Python function to create a logger with a custom log message format that includes the timestamp, logging level, and message.
2. Modify the function to use different formats for the file and console handlers.


```python
# 1. Write a Python function to create a logger with a custom log message format that includes the timestamp, logging level, and message.
import logging
def log_messages():
    logger = logging.getLogger("my_logger")
    logger.setLevel(logging.DEBUG)

    filehandler = logging.FileHandler('custom_app.log','w')
    file_format = logging.Formatter('%(asctime)s-%(levelname)s-%(message)s')
    
    filehandler.setFormatter(file_format)
    logger.addHandler(filehandler) 
    
    logger.debug("this is my custom logger")    
log_messages()
```


### Assignment 4: Rotating Log Files

1. Write a Python function to create a logger that uses a rotating file handler, which creates a new log file when the current log file reaches a certain size.
2. Modify the function to keep a specified number of backup log files.


```python
import logging
```

```python
from logging.handlers import RotatingFileHandler

def Rotating_file_log():
    logger = logging.getLogger("My_rotating_logger")
    logger.setLevel(logging.DEBUG)
    
    rotating_filehandler = RotatingFileHandler('111111rotating_file.log',maxBytes=2000, backupCount=5)
    rotating_filehandler.setLevel(logging.DEBUG)
    
    formatter = logging.Formatter('%(asctime)s-%(name)s-%(levelname)s-%(message)s')
    RotatingFileHandler.setFormatter(self=RotatingFileHandler,fmt=formatter)
    
    logger.addHandler(rotating_filehandler)
    
    for i in range(50):
        logger.debug('This message {}'.format(i))
        
Rotating_file_log()
    
```


### Assignment 5: Logging Exceptions

1. Write a Python function that logs an exception stack trace to a log file when an exception occurs.
2. Modify the function to log the stack trace at the ERROR level.


```python
def exception_log(a,b):
    logger = logging.getLogger("exception_logger")
    logger.setLevel(logging.DEBUG)
    
    file_handler = logging.FileHandler('exception_log.log')
    file_handler.setLevel(logging.DEBUG)
    
    formatter = logging.Formatter('%(asctime)s-%(name)s-%(levelname)s-%(message)s')
    file_handler.setFormatter(formatter)
    
    logger.addHandler(file_handler)
    
    try:
        result = a/b
        logger.debug("Division performed successfully bewteen {} and {} and the result is {}".format(a,b,result))
    except Exception as e:
        logger.exception("An Exception has occured")

exception_log(3,0)
    
```
