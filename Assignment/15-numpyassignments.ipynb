```python
## 1. Create two NumPy arrays of shape (3, 4) filled with random integers.
# Perform element-wise addition, subtraction, multiplication, and division.

array1 = np.random.randint(1,20,(3,4))
array2 = np.random.randint(21,40,(3,4))


print('Array1:\n',array1)
print('Array2:\n',array2)

print('-----Element wise Arthmatic Operation-------')

print('Addition')
print(array1 + array2)

print('Subtraction')
print(array1 - array2)

print('Multiplication')
print(array1 * array2)

print('Division')
print(array1 / array2)
```


### Assignment 5: Broadcasting

1. Create a NumPy array of shape (3, 3) filled with random integers. Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.
2. Create a NumPy array of shape (4, 4) filled with random integers. Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.


```python
## 1. Create a NumPy array of shape (3, 3) filled with random integers. Add a 1D array of shape (3,) to each row of 
#     the 2D array using broadcasting.


a = np.random.randint(1,9,(3,3))
print('First array shape is: 3x3')
print(a)
b = np.random.randint(1,10,3)
print('Second array shape is: 1x3')
print(b)

## broadcasting 
print('Third array shape is:3x3')
print(a+b)
```

```python
## 2. Create a NumPy array of shape (4, 4) filled with random integers. Subtract a 1D array of shape (4,)
#     from each column of the 2D array using broadcasting.


array1 = np.random.randint(1,10,(4,4))
array2 = np.random.randint(1,10,4)

print('First array of shape: ',array1.shape)
print(array1)


print('Second array of shape: ',array2.shape)
print(array2)

broatcast_array = array1 + array2
print('Broadcsted array',broatcast_array.shape)
print(broatcast_array)
```


### Assignment 6: Linear Algebra

1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.
2. Create two NumPy arrays of shape (2, 3) and (3, 2). Perform matrix multiplication on these arrays.


```python
## 1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.
array = np.random.randint(1,20,(3,3))
print('Matrix')
print(array)


print('\nDeterminent:\n',np.linalg.det(array))
print('\nInverse:\n',np.linalg.inv(array))
print('\nEigenvalues:\n',np.linalg.eigvals(array))
```

### Assignment 8: Fancy Indexing and Boolean Indexing

1. Create a NumPy array of shape (5, 5) filled with random integers. Use fancy indexing to extract the elements at the corners of the array.
2. Create a NumPy array of shape (4, 4) filled with random integers. Use boolean indexing to set all elements greater than 10 to 10.


```python
# 1. Create a NumPy array of shape (5, 5) filled with random integers. 
#    Use fancy indexing to extract the elements at the corners of the array.

arr = np.random.randint(1,10,size=(5,5))
print(arr)

## Fancy indexing
print('Fancy Indexing')
print(arr[[0,0,-1,-1],[0,-1,0,-1]])
```

```python
# 2. Create a NumPy array of shape (4, 4) filled with random integers. 
#     Use boolean indexing to set all elements greater than 10 to 10.

array = np.random.randint(1,50,(4,4))
print(array)

print('Boolean Indexing')
array[array>10]=10

print('Bool Indexed Array')
print(array)
```

### Assignment 10: Masked Arrays

1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.
2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

```python
# 1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. 
#     Compute the sum of the unmasked elements.

array = np.random.randint(1,17,(4,4))
print('Original Array')
print(array)

print('Masked Array')
print(array[array>10])
```

```python
# 2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. 
#    Replace the masked elements with the mean of the unmasked elements.

masked_array  = np.random.randint(1,10,(3,3))
print('Original Array')
print(masked_array)

diagnol_mask = masked_array.diagonal()
print(diagnol_mask)

# replace the masked element with the mean of unmasked element
print('Mean of Array:',np.mean(masked_array))
masked_array[[0,1,2],[0,1,2]] =np.mean(masked_array)

print(masked_array) 
```
