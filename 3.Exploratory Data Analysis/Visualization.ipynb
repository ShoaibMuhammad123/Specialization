# Seaborn for EDA

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## **Relational Plot**
1. Axes level = Scatter plot, line plot
2. figure level = relplot 

## **Distribution plots**
Distribution plots in seaborn show how a numeric variable (or variables) is distributed across its range.

`Figure-level function`

* displot() → (kind=: hist, kde, ecdf)

`Axes-level functions`

* histplot()

* kdeplot()

* ecdfplot()

## **Categorical plots**
categorical plots in seaborn visualize the distribution and/or relationship of variables where one or more variables are categorical.

`Figure-level function`

* `catplot()` → (kind=: strip, swarm, box, violin, boxen, point, bar, count)

`Axes-level functions`

* stripplot()

* swarmplot()

* boxplot()

* violinplot()

* boxenplot()

* pointplot()

* barplot()

* countplot()

# Figure level function Advantage
    Automatically creates and manages figure and multiple axes

    Built-in support for faceting using row, col, and hue

    Requires less code (minimal boilerplate)

    Provides consistent layout, scales, and legends across subplots

    Allows easy global customization of all subplots at once

    Ideal for quick exploratory data analysis (EDA) with meaningful defaults

* `Relational plots` (relplot)
1. → Used to show relationships between variables
2. → Examples: scatter plot, line plot

* `Distribution plots` (displot)
1. → Used to show distribution of data
2. → Examples: histogram, KDE, ECDF

* `Categorical plots` (catplot)
1. → Used to show relationship between categorical and numerical data
2. → Examples: box plot, violin plot, bar plot, strip plot
