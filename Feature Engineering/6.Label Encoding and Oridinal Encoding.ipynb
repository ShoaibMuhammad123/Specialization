## **Label Encoding and Ordinal Encoding**

### `Label Encoding`
* Label encoding and ordinal encoding are two techniques used to encode categorical data as numerical data.

* Label encoding involves assigning a unique numerical label to each category in the variable. The labels are usually assigned in alphabetical order or based on the frequency of the categories. `For example` if we have a category `"colors"` with three possible values "Red", "Green", "Blue". we can assign label in this way.

1. Blue = 1
2. Green = 2
3. Red = 3 

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "colors":['Red','Green','Red','Blue','Blue','Green','Blue','Red','Green']
})
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lb_encoder = LabelEncoder()
lb_encoder.fit_transform(df[['colors']])
```

```python
lb_encoder.transform([['Red']])
```

```python
lb_encoder.transform([['Green']])
```

```python
lb_encoder.transform([['Blue']])
```

```python
## Encoding the label 

df['colors_label_encode'] = lb_encoder.transform(df['colors'])
```

```python
df
```

`Problem with Label Encoding: `
* If the categories has some order, at that time we can't use `label encoder` because it is not a good techniques for that time.
`Solution:`
* To encoder the ordinal data , mean if the data has some order like T-shirt size can be `small` -> `Medium`-> `Large`. We can use `Ordinal Encoding.`

### **Ordinal Encoding**

In ordinal Encoding we assign highest value to the high rank category and lowest to the low rank category. like `small`=1, `medium`=2, and `large`=3


```python
df1 = pd.DataFrame({
    'Education Level':['M','FSC','BS','MS','M','PhD','PhD']
})

df1
```

```python
from sklearn.preprocessing import OrdinalEncoder

or_encoder = OrdinalEncoder(categories=[['M','FSC','BS','MS','PhD']])
```

```python
encoded = or_encoder.fit_transform(df1[['Education Level']])
```

```python
encoded
```

```python
encoded_df = pd.DataFrame(
    encoded,
    columns=or_encoder.get_feature_names_out(['Education Level']),
    index=df1.index
)
```

```python
encoded_df
```

That is how we can apply Ordinal Encoding 

```python
df1['Education ordinal Encode'] = or_encoder.transform(df1[['Education Level']])
```

```python
df1
```

```python
new_df = pd.concat([df1,encoded_df],axis=1)
```
