## **Target Guided Encoding in Ordinal Way**
* It is a technique used to encode `categorical variables` based on their relationship with the target variable. This encoding techniques is useful when we have a categorical variable with `large number of unique categories`, and we want to use this variable as a feature in our machine learning model.

* In Target Ordinal Encoding we replace each category in the categorical variable with a numerical value based on the mean or median of the target variable for that category. This create a monotonic relationship between the categorical variable and the target variable, which can improve the predictive power of our model.

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Education':['M','Phd','Phd','MS','BS','M','MS','BS'],
    'salary':[100,1000,1200,700,300,250,500,400]
})
```

```python
df
```

`Target Guided Encoding`

```python
## first of all we group the data
mean_salary = df.groupby('Education')['salary'].mean().to_dict()
mean_salary
```

```python
df['Education_encode'] = df['Education'].map(mean_salary)
```

```python
df
```

#### Now I can give only `two feature` to my model `for training purpose`

```python
# the two features are 
df[['salary','Education_encode']]
```

```python
import seaborn as sns
```

```python
tips_df = sns.load_dataset('tips')
```

```python
tips_df
```

```python
## Target Guided Ordinal encoding of Time with total bill

mean_total_bill = tips_df.groupby('time')['total_bill'].mean().to_dict()
```

```python
mean_total_bill
```

```python
tips_df['time_encode_guided_oridinal'] = tips_df['time'].map(mean_total_bill)
```
