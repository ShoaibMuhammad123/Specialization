## **Handling Imbalanced Dataset**
There are two techniques of Handling imbalanced dataset. 
* **Up Sampling** -> Here we increase the less class datapoints.
* **Down Sampling** -> Here we decrease the high class datapoints.

```python
import pandas as pd 
import numpy as np
```

```python
## Selecting random seed for reproducibility
np.random.seed(123)

# Create a dataframe with two classses

n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples*class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

`Creating my DataFrame with Imbalanced Data`

```python
class_0 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0] * n_class_0
})
class_1 = pd.DataFrame({
    'feature_1':np.random.normal(loc=2,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=2,scale=1,size=n_class_1),
    'target':[1] * n_class_1
})
```

```python
class_0
```

```python
class_1
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.target.value_counts()
```

### `upsampling` 
* Increasing the less number of class samples

```python
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample

df_minority_upsampled = resample(df_minority,replace=True, # sample with replacement
                        n_samples=len(df_majority),
                        random_state=42
                        )
```

```python
df_minority.shape
```

```python
df_minority_upsampled.shape  # this is how we upsample the minority class
```

```python
df_minority_upsampled.head()
```

```python
## Now we can combine majority with minority 

df_upsampled = pd.concat([df_majority,df_minority_upsampled])
```

```python
df_upsampled.shape
```
