```python
# speed difference 

# list

a= [i for i in range(10000000)]
b= [i for i in range(10000000,20000000)]


c=[]
import time

start = time.time()

for i in range(len(a)):
    c.append(a[i]+b[i])
print(time.time()-start)
```

## Broadcasting

#### the term broadcasting describes how Numpy treats arrays with different shapes during arithmetic operaations

```python
# Broading tell than how numpy treat with different shape of arrays while performing arthimatic operation
# The smaller array is "broadcasted " across the larger array so that they have compatible shapes
```

```python
# same shape

a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)

# this is item wise addition

print(a+b)
```

```python
# differ shape

a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)
print(a)


print(b)
# there question arise that how 2 differnt shape arrays can be perform addittion on them
print(a+b)
```

```python
# it is only perform through broadcasting
```

### Broadcasting rules

#### 1- Make the two array have the same number of dimentions
###### If the numbers of dimentions of the two arrays are different add nw dimensions with size 1 to the head of the array with the samller dimension
#### 2 Make each dimension of the two arrays the same size

###### if the sizes of each dimension of the two arrays do not match dimensions with size 1 are stretched to the size of the other array

###### if there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted. and an error is raised

```python
# broadcasting examples 

a = np.arange(12).reshape(4,3)
b = np.arange(3)
print(a )
print(b)
print('now this will performed broadcasting')
print(a+b)
```

```python
 # this give you an error
    
    
    
a = np.arange(12).reshape(3,4)
b = np.arange(3)


print('this give you an error',a+b)
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(3).reshape(3,1)
print(a+b)
```

```python
# shape --> (1,1)
# shape --> (2,2)

a = np.array([1])
b = np.arange(4).reshape(2,2)

print(a+b)
```

```python
# broadcasting can't performed because 1 not exist

a = np.arange(12).reshape(3,4)
b = np.arange(12,24).reshape(4,3)


print(a+b)
```
