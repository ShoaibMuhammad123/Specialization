## np.concatinate()

```python
# numpy.concatenate() function concatenate a sequence of arrays along an existing axis
```

```python
c= np.arange(6).reshape(2,3)
d = np.arange(6,12).reshape(2,3)

print(c)
print(d)


np.concatenate((c,d),axis=0)
```

```python
np.concatenate((c,d),axis=1)
```

```python
f = np.arange(20).reshape(2,10)
g = np.arange(40,60).reshape(2,10)

print(f)
print(g)

np.concatenate((f,g),axis = 0)
```

```python
np.concatenate((f,g),axis =1 )
```

## np.where()
### it return the indices of elements in an input array where the given condition is satisfied

```python
p
```

```python
# if we need the index position of those elements which are > 3 (any condition)
np.where(p>3)
```

```python
a=np.random.randint(1,100,15)
a
```

```python
np.where(a>50)   # it give the index of those elemnets which satisfy the condition
```

```python
# Another use of this function to replace some values that satisfiy condition or not satisfy np.where(condtion,true,false)


np.where(a>50,0,a)
```

```python
np.where(a<50,a,1)
```

```python
# replace even number with 0

np.where(a%2==0,0,a)
```
