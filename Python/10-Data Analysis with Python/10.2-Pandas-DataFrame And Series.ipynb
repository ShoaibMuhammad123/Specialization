## `Pandas DataFrame And Series`

* Pandas is a powerful data manipulation library in Python, widely used for `data analysis` and `data cleaning`. It provides two primary data structures: `Series` and `DataFrame`. A series is a one dimentional array like object, while a DataFrame is a two dimentional, size-mutable and potentially heterogenous tabular data structure with labeled axes (rows and columns)

`data manipulation: ` 
* mean modifying and organizing data to make it useful.
* It includes tasks like `filtering, sorting, grouping, cleaning, and transforming data`.

`Size-mutable`
* Size-mutable means a DataFrame can grow or shrink.
You can add or remove rows and columns after it is created.

`Potentially heterogeneous`
* Potentially heterogeneous means a DataFrame can contain different data types.
* For example, one column can have numbers, another text, and another dates.

```python
import pandas as pd
import numpy as np
```

`Series`
* Series is a one dimentional array like object that can hold any data type. It is similar to a column in a table.

```python
data = [1,2,3,4,5]

series = pd.Series(data)
print('Series\n',series)
print(type(series))
```

* `Creating series from dictionary element`

```python
data = {'a':1,'b':2,'c':3}
series_dict = pd.Series(data)   # here the keys will be considered as index and values will be value in series
print(series_dict)
```

* `We can also give different index over here`

```python
## another technique to create a series

data = [10,20,30,40]

index = ['A','B','C','D']   
pd.Series(data,index)
```

## `DataFrame`
* In data frame we can have multiple `rows` and `columns`

```python
## Creating a dataframe from a dictionary of lists

dict_of_list = {
    'shoaib':[12,34,56,77,98],
    'Alim':[44,55,23,41,21],
    'Kamil':[44,21,22,13,63]
}

print(dict_of_list)

pd.DataFrame(dict_of_list)
```

```python
data = {
    'Name':['Shoaib','Kamil','Qais','Kashif'],
    'Age':[33,44,53,24],
    'City':['Peshawar','Karachi','Islamabad','Quetta']
}


df = pd.DataFrame(data)
print(type(df))
df
```

```python
## now you can convert this into an array also
np.array(df)
```

```python
## Creating a dataframe from a list of Dictionaries
data = [
    {'Name':'Shoaib','Age':32,'City':'KPK'},
    {'Name':'Amin','Age':44,'City':'Islamabad'},
    {'Name':'Kashif','Age':35,'City':'Quetta'}
]
df1 = pd.DataFrame(data)
print(type(df1))
df1
```
