### `Reading Data`

```python
import pandas as pd
from io import StringIO
```

* If you have a `key- value` pairs then you can converted that into a `dataframe`

```python
data = '{"employee_name":"shoaib","email":"shoaib@gmail.com","job_profile":[{"title_1":"team_lead","title_2":"Sr.Developer","title_3":"AI Engineer"}]}'

df = pd.read_json(StringIO(data))
df
```

* If I want to convert the `dataframe` back into a `json`

```python
df.to_json()
```

* but above you can see the conversion into json give some issue because it give some else key 

```python
## to convert into json based on records we pass a argument to "orients=records" by default it is orient =index

df.to_json(orient='records')  # now it give us correct anser
```
