1- `Handling Missing Values`

```python
df.isnull().sum()
```

* To check the missing value `with respect to every record` we do 

```python
df.isnull().any(axis=1)  # It return "True " if any record has a missing value and "False" if record has no missing value
```

```python
df.isnull().any()  # by default the axis=0, it check for each column
```

i-` fill all the missing with 0`

```python
## to fill all the missing with 0
df_filled = df.fillna(0)
```

ii-` fill all the missing with mean of the columns`

```python
df['avg_rating_fillNA_mean'] = df['avg_rating'].fillna(df['avg_rating'].mean())
```

```python
df.head(3)
```

iii-`Changing type of column`

```python
df.dtypes
```

```python
### Renaming column
df = df.rename(columns={'model':"Phone_Model"})
```

```python
df.head()
```

```python
## Changing the data type 

df['Value_new'] = df['avg_rating'].astype('float16')
```

```python
df.dtypes
```

```python
df['New_value'] = df['Value_new'].apply(lambda x:x*2)
```

```python
df.head(3)
```

3- `Data Aggregating and Grouping`

* Based on the `os (operating system)` i want to find the `average rating of each brand`

```python
## if we want to aggregate
grouped_avg_rating = df.groupby('os')['avg_rating'].mean()

print(grouped_avg_rating)
```

3.1 - `If I have to group by based on two categories`

```python
df.groupby(['brand_name','os'])['avg_rating'].mean().head(5)
```

```python
df.groupby(['brand_name','os'])['price'].sum().head(10)
```

4- `Multiple Aggregate functions`

```python
df.groupby(['processor_brand','os'])['price'].agg(['mean','sum','max','min','count'])
```

5-`Merging and Joining DataFrame`

```python
## Sample data frame
df1 = pd.DataFrame({'Key':['A','B','C'],'Value':[1,2,3]})
df2 = pd.DataFrame({'Key':['A','B','D'],'Value':[4,5,6]})
```

```python
df1
```

```python
df2
```

4.1 - Merging dataframe on basis of `key column`

* how = `inner` : mean that only matching column need to be kept

![alt text](images.png)

![alt text](join-types-merge-names.jpg)

```python
pd.merge(df1,df2,on = 'Key',how='inner')
```

* how =`Outer or Full`

```python
pd.merge(df1,df2,on = 'Key',how='outer')
```

* `left` dataframe is taken overall and from right only the matching will be kept

```python
pd.merge(df1,df2,on = 'Key',how='left')  
```

* `right` right is taken overall and left only matching will be keep

```python
pd.merge(df1,df2,on = 'Key',how='right')  
```
