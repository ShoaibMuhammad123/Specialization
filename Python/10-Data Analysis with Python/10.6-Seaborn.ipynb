##### `Data Visualization with Seaborn`
* Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code. In this lesson, we will cover the basics of Seaborn, including creating various types of plots and customizing them.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
## Basic Plotting with Seaborn

tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
## 1- Creating Scatter plot
sns.scatterplot(x ='total_bill',y ='tip',data=tips)
```

```python
## 2-Creating line plot
sns.lineplot(data =tips,x='size',y='total_bill')
plt.title('Size Vs Tips')
```

```python
## Categorical Variable
## 3-Bar plot
sns.barplot(data=tips,x='day',y='total_bill')
```

```python
## 4- Box Plot
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
## 5-Violin Plot
sns.violinplot(data=tips,x='day',y='total_bill')
```

```python
### 6- Histogram
sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
## 7-KDE plot
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
## 8-Pair Plot
# Pair plot is used if i want to see the relationship of all the variable wrt each other

sns.pairplot(tips)
```

```python
corr = tips.corr(numeric_only=True)
corr
```

```python
## 9-Heatmap
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
df = pd.read_csv('smartphones.csv')
df.head(3)
```

```python
## plot total sales for each brand
plt.figure(figsize=(10,5))
sns.barplot(data=df,y='processor_brand',x='price',estimator=sum)
plt.xlabel('Total Price')
plt.xlabel('Processor Brands')
plt.title('Total Price for each Process brand combine phone')
```

```python
## plot total sales for each brand
plt.figure(figsize=(10,5))
sns.barplot(data=df,y='price',x='os',estimator=sum)
plt.xlabel('Total Price')
plt.xlabel('Processor Brands')
plt.title('Total Price for each Process brand combine phone')
```
