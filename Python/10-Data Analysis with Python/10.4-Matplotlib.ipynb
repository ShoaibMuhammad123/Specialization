### `Data visualization with Matplotlib`

* Matplotlib is a powerful plotting library for python that enables the creation of static, animted and inteeractive visualization. It is widely used for data visualization in data science and analytics. here we cover basics of matplotlib, including various types of plots and customizing them.

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [22,33,37,42,46]

## to create a line plot
plt.plot(x,y)
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title("Basic Line plot")
```

* Customized line plot

```python
x = [1,2,3,4,5]
y = [22,33,37,42,46]

plt.plot(x,y,color='red',linestyle='--',marker='o',linewidth=3)
plt.grid(True)
```

`Multiplot`
* To create multiplot we use `Subplots`

```python
x = [1,2,3,4,5]
y1 = [22,44,55,66,77]
y2= [1,2,3,4,5]

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y1,color='green')
plt.title('plot-1')

plt.subplot(2,2,2)
plt.plot(y1,x,color='red')
plt.title('plot-2')

plt.subplot(2,2,3)
plt.plot(x,y2,color='teal')
plt.title('plot-2')


plt.subplot(2,2,4)
plt.plot(y2,x,color='red')
plt.title('plot-2')
```

2-`Bar plot`

```python
## bar plot
categories = ['A','B','C','D','E']

values = [5,7,2,8,9]

plt.bar(categories,values,color='brown')
```

`Histogram`
* Histogram are used to represent the distribution of dataset. They divide the data into bins and count the number of data points in each bin

```python
data = [1,2,2,3,4,5,6,6,6,7,7,7,8,9,9,4]

## create a histogram
plt.hist(data,bins=5,color='orange',edgecolor='black')   # we are saying that take this data and dive into 5 bins or categries
```

`Scatter plot`

```python
x = [1,2,3,4,5,6]
y = [2,3,4,5,6,7]

plt.scatter(x,y,marker='x')
```

`Pie chart`


```python

label = ['a','b','c','d']
sizes = [20,30,40,50]

colors =['b','r','y','g']
explode = [0.2,0,0,0]
plt.pie(sizes, labels=label, explode=explode,colors=colors,  autopct='%1.1f%%', shadow=True)



```

### `Practical Example`

```python
import pandas as pd
df = pd.read_csv('smartphones.csv')
```

```python
df.head(3)
```

```python
## plot total sales by brand name
total_sales_on_brand_name = df.groupby('brand_name')['price'].sum().sort_values(ascending=False).head(10)
```

```python
total_sales_on_brand_name
```

```python
total_sales_on_brand_name.plot(kind='bar',color='red')
```

```python
os_processor_df = df.groupby(['os','processor_brand'])['price'].sum().reset_index()
```

```python
os_processor_df.head()
```

```python
plt.bar(os_processor_df['os'],os_processor_df['price'])
```

```python
plt.barh(os_processor_df['processor_brand'],os_processor_df['price'])
```
