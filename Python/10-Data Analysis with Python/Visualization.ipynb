* `Relative Frequency Distribution Table`

* Cumulative frequency

```python
freq = planets['year'].value_counts()

freq.cumsum().sort_index(ascending=True).plot.line()

```

```python
sns.lineplot(planets['year'].value_counts().cumsum(),color='red')
```
