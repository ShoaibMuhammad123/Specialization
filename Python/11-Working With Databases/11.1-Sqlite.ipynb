### `SQL AND SQLITE`
* SQL(Structured Query Language) is a standard language for managing and manipulating relational databases. `SQLITE is a self-contained, serverless, and zero-configuration databases engine that is widely used for embedded database systems.` In this lesson, we will cover the basics of SQL  and SQLite , including creating databases, tables , and performing various SQL operation. 
* SQLite is a `light version` of SQL

```python
import sqlite3
```

###  First we need to `create a Sqlite database`

```python
## connect to an SQLite database
connection = sqlite3.connect('example.db')   ## here we create a database in our local system
connection
```

* Now we create a `cursor object :(throught which we iterate all the table or rows that's why we create a cursor)`

```python
## creating cursor object
cursor = connection.cursor()
```

* To create a table in the database we use  `cursor object` with `execute()` function and inside the execute function `what ever my command is will be written in the execute function `

```python
## Create a table 

## to create a table we use the same curor object
cursor.execute('''
    Create Table If Not Exists employees(
        id Integer Primary Key,
        name Text Not Null,
        age Integer,
        department text
    )
    ''')

## When it execute we need to commit the changes
connection.commit()
```

```python
## It has no data because the table is just create 
cursor.execute('''
    Select * from employees
    ''')
```

### `Inserting` data in Database

```python
## Insert some data in sqlite table
cursor.execute('''
    Insert Into employees(name,age,department)
                values('Shoaib',22,'Student')

''')

## to insert some more record 
cursor.execute('''
    Insert Into employees(name,age,department)
                values('Akram',25,'Teacher')

''')

cursor.execute('''
    Insert Into employees(name,age,department)
                values('Muhsina',29,'Data Scientist')

''')


## now we commit the changes
connection.commit()
```

```python
## to access / retrive data from database table 
cursor.execute('Select * from employees')     ## once we execute it we get the rows

rows = cursor.fetchall()  ## to get all the rows


## to print all the rows data 
for row in rows:
    print(row)
```

### `Update` the data into a table

```python
## Update data in table 
cursor.execute('''
    Update employees
    Set age=25
    where name ='Shoaib'
         ''')

## commit changes
connection.commit()
```

```python
cursor.execute('Select * from employees')

rows = cursor.fetchall()

for row in rows:
    print(row)

```

### `Delete` the data from the table

```python
cursor.execute('''
    
    Delete from employees
        where name='Akram'   
    ''')

## commit changes
connection.commit()
```

```python
cursor.execute('Select * from employees')

rows = cursor.fetchall()

for row in rows:
    print(row)
```

### `Working on Sales Data Project`

```python
## first we connect to a sqlite database
connection = sqlite3.connect('sales_data.db')
## to interact with database table and  rows we create a cursor object
cursor = connection.cursor()

## now we create a table in sales database by the name of sales
cursor.execute('''
    Create Table If NOT Exists sales(
        id INTEGER PRIMARY KEY,
        date TEXT NOT NULL,
        product TEXT NOT NULL,
        sales INTEGER,
        region TEXT
    )           
    ''')

## Inserting data into sales table
sales_data =[
    ('2023-01-01','Product',100,'North'),
    ('2023-01-02','Product',200,'North'),
    ('2023-01-03','Product',150,'North'),
    ('2023-01-04','Product',250,'North'),
    ('2023-01-05','Product',300,'North')
]

## here we are inserting many record so, we use cursor.executemany
cursor.executemany('''
        Insert Into sales(date,product,sales,region)
                    values(?,?,?,?)       
        ''',sales_data)

## commit changes 
connection.commit()
```

```python
cursor.execute('Select * from sales')

rows = cursor.fetchall()

for row in rows:
    print(row)
```

* to Permanently close the connection

```python
## close the connection
connection.close()
```

```python
## now if we try to fetch data it gives us an error because connection has been closed
```

## MY on practice

```python
connection =  sqlite3.connect('My_database.db')
cursor = connection.cursor()

## table create in db
cursor.execute('''
        Create Table If Not Exists student(
            id Integer PRIMARY KEY,
            name TEXT NOT NULL,
            age INTEGER NOT NULL,
            gender TEXT NOT NULL,
            marks INTEGER NOT NULL
            )       
        ''')

## add information in table
cursor.execute('''
        Insert Into student(name,age,gender,marks)
                    values('Shoaib',22,'male',82)       
        ''')

cursor.execute('''
        Insert Into student(name,age,gender,marks)
                    values('kumar',23,'male',88)       
        ''')

cursor.execute('''
        Insert Into student(name,age,gender,marks)
                    values('salma',22,'female',90)       
        ''')

connection.commit()
```

```python
## RETRIEVE DATA FROM DATABASE
cursor.execute('Select * from student')

records = cursor.fetchall()
for record in records:
    print(record) 
```

```python
## Update Information in Database Table

cursor.execute('''
        
        UPDATE student
        Set age = 30 
        Where name='kumar'         
        ''')

connection.commit()

```

```python
cursor.execute('SELECT * FROM student')

updated_record = cursor.fetchall()

for record in updated_record:
    print(record)
```
