### `Logging with Multiple Loggers`
You can create multiple loggers for different parts of your application.

```python
import logging 
## create a logger for module1

# inorder to create a logger for a specific module we use logger.getLogger("module name")
logger1 = logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)

## create a logger for module 2
logger2 = logging.getLogger('module2')
logger2.setLevel(logging.WARNING)

## Configure logging setting
logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(name)s - %(levelname)s - %(message)s",
    datefmt='%Y-%m-%d %H:%M:%S'
)
```

```python
## now i will go and log messages (so what ever logger you want you can use it)
logger1.debug("This is a debug message")

logger2.warning("This is a warning message")
logger2.error("This is a error message")
```

* In previous logging the default module was `root` but over here we define that for each module
