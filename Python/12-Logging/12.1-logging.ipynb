### `Python Logging`
* Logging is a crucial aspect of any application, providing a way to track events, errors, and operational information. Python's built-in logging module offers a flexible framework for emitting log messages from Python programs. In this lesson, we will cover the basics of logging, including how to configure logging, log levels, and best practices for using logging in Python applications.

```python
import logging
import time

## after import logging we need to configure the basic logging setting
logging.basicConfig(level=logging.WARNING)

## log messages with different severity levels
logging.debug("This is the debug message")
logging.info("This is a info message")
logging.warning("This is a warning message")
logging.error("This is a error message")
logging.critical("This is the critical message")

```

### `Log Levels`
* DEBUG: Detailed information, typically of interest only when diagnosing problems. 
* INFO: confirmation that things are working as expected. 
* WARNING: An indication that something unexpected happened or indicative of some problem near in future( e.g disk space low). The softwarre is still working as expected.
* ERROR: Due to a more serious problem, the software has not been able to perform some function.
* CRITICAL: A very serious error, indicating that the program itself may be unable to continue running. 

Here `severity means` how serious or important a log message is.

In Python **logging**, the severity level tells you the seriousness of the situation that the message describes:

`DEBUG` → very low severity (details for developers)

`INFO` → normal information

`WARNING` → something unexpected, but program still works

`ERROR` → a serious problem occurred

`CRITICAL` → very serious problem, program may stop

```python
## Configuring the logging
logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s- %(name)s- %(levelname)s- %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S"
)


## log messages with different severity levels
logging.debug("This is the debug message")
logging.info("This is a info message")
logging.warning("This is a warning message")
logging.error("This is a error message")
logging.critical("This is the critical message")
```

* `To save the logging in a file we need to store that for future use`


```python
import logging
import time

logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format="%(asctime)s-%(name)s-%(levelname)s-%(message)s",
    datefmt='%Y-%m-%d %H:%M:%S'
)

logging.debug("This is the debug message")
logging.info("This is a info message")
logging.warning("This is a warning message")
logging.error("This is a error message")
logging.critical("This is the critical message")
```
