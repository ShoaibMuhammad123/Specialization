## `Iterators`
* Iterators are advanced python concepts that allow for efficient looping and memory management. Iteration provide a way to access elements of a collection sequentially with out exposing the underlying structure.

```python
my_list = [1,2,3,4,5,6]
for i in my_list:
    print(i)
```

```python
## we we check the type 
type(my_list)
```

```python
print(my_list)
```

`How we create an Iterator`
* To create an iterator we will be using an `iter()` keyword and pass the iterable.

```python
iterator = iter(my_list)  ## it gives us an iterator
iterator
```

```python
## we we check the type of an iterator 

print(type(iterator))
```

* When we create an iterator it directrly not gets the element
* Because it is using a `lazy learning techniques`
* It will not give us an element unless and untill we iterate through that element.

```python
## Inorder to iterate through all the element we use "next()" function

next(iterator)    # when we execute it return a single element
```

* when the element gets complete it show an exception  `StopIteration`

```python
next(iterator) 
```

* To handle that `StopIteration` Exception

```python
# we can use try except block
try:
    print(next(iterator))
    
except StopIteration:
    print('There is no element left for iteration')
    
```

```python
## You can also make iterator for string also
text = "Hello"
str_iterator = iter(text)

print(next(str_iterator))
print(next(str_iterator))
print(next(str_iterator))
print(next(str_iterator))
print(next(str_iterator))
```

`Why we are using this:` 
* It is for efficient looping and Memory Management
