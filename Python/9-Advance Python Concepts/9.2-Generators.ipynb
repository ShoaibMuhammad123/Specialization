## `Generators`
* Generators are a simpler way to create iterators. They use the `yield` keyword to produce a series a value `Lazily` which mean they generate values on the fly and do not store them in memory

```python
def square(n):
    for i in range(n):
        return i**2
```

```python
square(3)  ## here it return 0
```

```python
# but what we actually want is we want to get the square all the 
# element in the iteration

# to do this we use generator with yeild keyword

def square(n):
    for i in range(n):
        yield i**2
        
gen = square(3)
```

```python
next(gen)
```

```python
gen   # generator are simple way of creating iterator
```

```python
for i in gen:
    print(i)
```

```python
### Another example for creating generator
def my_gen():
    yield 1
    yield 2
    yield 3
    yield 4
    
```

```python
g = my_gen()
g
```

```python
print(next(g))
print(next(g))
print(next(g))
```

```python
## use for loop
for i in g:
    print(i)
```

#### `Practical Examples`
#### Reading Large Files
* Generator are particularly usefull for reading large files because they allow you to process one line at a time with out loading the entine file into memory

```python
def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            # print(line,end='')
            yield line
            

    
```

```python
for line in read_large_file('larefile.txt'):
    print(line.strip())
```

`Difference b/w Iterator and Generator`
* For  `Iterator` you have to use `iter()` keyword.
* For `Generator` you have to use `yield` keyword.
