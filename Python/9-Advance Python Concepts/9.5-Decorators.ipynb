`Decorators`
* Decorators are a powerful and flexible feature in python that allows you to modify the behavior of a function or class method. `They are commonly used to add functionality to functions or methods without modifying their actual code.` This lesson covers the basics of decorators, including how to create and use them.

Before we start decorator first we need to understand `function copy` and `closures` then we move to `Decorators`
* function copy
* closures
* decorators

```python
### function copy
def display(name):
    return f"my name is {name}"
```

```python
display("shoaib")  # calling function normally
```

```python
## function copy 

my = display  # assigning the same function to another variable 
      
my("shoaib")  # calling on the basis of the name

```

```python
# if we delete the assigned function then the other properly work

del display
```

```python
display() # now it is not showing anything because it is deleted
```

```python
my("shoaib")   
```

* This is known as `function copy`

`Clousers`
* It is nothing but a function inside another function
* in a clousers function any thing which is defined outside the submethod can be access in the submethod as well

```python
##  clousers functions 
def main_welcome(msg):
    def inner_welcome():
        print(msg)
    return inner_welcome()
        
main_welcome("Shoaib Welcome")
```

```python
def main_welcome():
    msg = "Welcome message outside sub function"
    def inner_welcome():
        print("You are in the sub method")
        print(msg)
        
        
    return inner_welcome()
        
main_welcome()
```

```python
# we can also pass the function to the function

def main(func):
    def inner_func():
        print('inside inner function')
        print(func())
        print('after calling the arg function')
    return inner_func()
def passing():
    return 'message of arg function'
    
main(passing)
```

### `Decorators`

```python
def main_welcome(func):
    
    def sub_welcome_method():
        print("Welcome to the advance python course")
        func()
        print("please learn these concepts ")
        
    return sub_welcome_method()

```

```python
def course_intro():
    print('this is advance python course')
    
course_intro()
```

```python
main_welcome(course_intro)
```

* Now can i without calling `main_welcome` function can i make sure i display all these messages, this is possible through  `Decorator`

```python
@main_welcome
def course_intro():
    print('this is advance python course')
```
