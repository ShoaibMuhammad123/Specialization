### `Generator`
* Python generators are a simple way of creating Iterators.

* Python generator is nothing but a function, that has just a little difference from the normal function.

```python
def gen_demo():
    yield "first statement"
    yield "second statement"
    yield "third statement"

```

```python
gen = gen_demo()
for i in gen:
    print(i)
```

* Another example

```python
def square(num):
    for i in range(num):
        yield i**2
        
```

```python
gen = square(10)

print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
for i in gen:
    print(i)
```

#### Range function using generator

```python
def mera_range(start,end):
    for i in range(start,end):
        yield i
```

```python
   # using generator range function my own
def mera_range(start,end):
    start =start
    end = end
    while True:
        if start >=end:
            break
        else:
            current = start
            start+=1
            yield current
        
```

```python
gen = mera_range(4,10)

for i in gen:
    print(i)
```

```python
for i in mera_range(1,10):   # working of my own mera_range()
    print(i)
```

`Generator expression`
* Making that more simple is generator expression 
* It is much similiear to list expression but i use () brakets

```python
## list comprehension
L = [i**2 for i in range(10)]
```

```python
gen = (i**2 for i in range(1,10))
print(gen)  # it is a generator
```

```python
# to access the generator element one by one 
for i in gen:
    print(i)
```
