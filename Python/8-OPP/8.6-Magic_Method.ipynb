## `Magic Method`

* In Python, magic methods (also called dunder methods, because they have double underscores before and after their names, like __init__ or __str__) are special methods that allow developers to define or customize the behavior of objects. They are automatically invoked by Python in response to built-in operations, enabling objects to interact seamlessly with the language’s syntax. For example, __init__ is used to initialize a new object when it is created, while __str__ or __repr__ defines how the object is represented as a string, such as when printed. Magic methods also allow objects to support arithmetic operations (__add__, __sub__), comparisons (__eq__, __lt__, __gt__), and container-like behavior (__len__, __getitem__, __setitem__), letting custom objects behave like numbers, lists, or dictionaries. Additionally, methods like __del__ can control the object’s lifecycle by running code when an object is about to be destroyed. The primary purpose of magic methods is to make objects integrate naturally with Python’s syntax and built-in functions, so that operations like addition, printing, indexing, or comparisons work intuitively without requiring separate function calls. By implementing magic methods, developers can create classes that feel and behave like native Python types, providing both flexibility and readability in their code.

```python
class Coin:
    def __init__(self,v2,v3):
        self.v2 = v2
        self.v3 = v3
        
        print('This is init magic method')
        
    def __add__(self,other):
        return self.v2 + other.v2 ,self.v3 + other.v3
    
    def __sub__(self,other):
        return self.v2 -other.v2 , self.v3 - other.v3
    
    def __mul__(self,other):
        return self.v2 * other.v2, self.v3 * other.v3
        
    # def __str__(self):
    #     return f"{self.v2},{self.v3}"
    def __repr__(self):
        return f"{self.v2},{self.v3}"
    

point1 = Coin(3,4)
point2 = Coin(4,2)

print(point1)

print(point1+point2)
print(point1-point2)
print(point1*point2)
```

## vector

```python
class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y
        
    def __add__(self,other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __sub__(self,other):
        return Vector(self.x - other.x,self.y - other.y) 
    
    
    def __gt__(self,other):
        return Vector(self.x - other.x,self.y - other.y) 
    
    def __eq__(self,other):
        return Vector(self.x == other.x and self.y == other.y) 
    
    
    def __repr__(self):
        return f"Vector({self.x},{self.y})"
    
v1 = Vector(2,3)
v2 = Vector(5,3)

print(v1+v2)
print(v1-v2)

print(v1<v2)
print(v1>v2)
```
